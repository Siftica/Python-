## 1. 从一个实际问题开始

上一课的词表是 `['我', '爱', '学', '习', '模', '型']`。查询已有字符会成功，但查询“大”会报错：

```python
vocab.index("大")  # 大不在词表里，会报 ValueError
```

这是文字中的报错示例，不会自动运行。本课采用一条新规则：

- 字符在词表里：使用它自己的编号。
- 字符不在词表里：使用事先约定的未知字符编号。

## 2. 字典 dict：保存键与值的对应关系

字典用花括号 `{}` 创建，每项写成 `键: 值`。例如 `"我": 0` 中，“我”是键，0 是值。

下面保留原来的 0–5 号，新增 `"<unk>": 6`。unk 来自 unknown，表示未知。

**先预测：** 查询“我”和“模”分别会得到什么？


````python
token_to_id = {
    "我": 0,
    "爱": 1,
    "学": 2,
    "习": 3,
    "模": 4,
    "型": 5,
    "<unk>": 6,
}

print(token_to_id["我"])
print(token_to_id["模"])
````

### 列表与字典的查找方式

| 表达式 | 意思 | 本课结果 |
| --- | --- | --- |
| `vocab[0]` | 按列表位置取值 | `"我"` |
| `token_to_id["我"]` | 按字典中的键查值 | `0` |

`token_to_id[0]` 不表示“取字典的第一个值”，而是查找数字 0 这个键。本课字典没有这个键，直接查询会报 KeyError。

`"<unk>"` 是我们手工约定的标记。Python 不会自动识别它的含义；稍后的 else 分支会决定何时使用它。

## 3. 用 in 判断键是否存在

`"我" in token_to_id` 检查“我”是不是这个字典里的键。

True 表示判断成立，False 表示不成立。这里检查的是键，不是值。


````python
print("我" in token_to_id)
print("大" in token_to_id)
````

## 4. if / else：根据条件选择一个分支

把下一单元读成：如果字符在字典里，就查它自己的编号；否则，使用未知标记的编号。

- if 与 else 所在行的末尾都有冒号。
- else 与对应的 if 对齐。
- 分支中的语句缩进 4 个空格。
- 一次这样的判断只会执行两个分支中的一个。

**先预测：** char 是“大”时，最后打印几？把它改成“模”又会打印几？


````python
char = "大"

if char in token_to_id:
    token_id = token_to_id[char]
else:
    token_id = token_to_id["<unk>"]

print(token_id)
````

## 5. 与循环组合

下一单元包含完整词表，可以独立运行。循环内部逐字符选择编号，再追加到列表。

为了观察执行过程，循环中会打印“当前字符、选择的编号、追加后的列表”。


````python
token_to_id = {
    "我": 0,
    "爱": 1,
    "学": 2,
    "习": 3,
    "模": 4,
    "型": 5,
    "<unk>": 6,
}
text = "我爱大模型"
token_ids = []

for char in text:
    if char in token_to_id:
        token_id = token_to_id[char]
    else:
        token_id = token_to_id["<unk>"]

    token_ids.append(token_id)
    print(char, token_id, token_ids)

print("完整编号：", token_ids)
````

### 5.1 理解 append 的位置

| 语句 | 缩进位置 | 作用 |
| --- | --- | --- |
| `for char in text:` | 最外层 | 按顺序处理字符 |
| `if ...` / `else:` | 4 个空格 | 为当前字符选择分支 |
| `token_id = ...` | 8 个空格 | 在被选择的分支里确定编号 |
| `token_ids.append(token_id)` | 4 个空格 | 每个字符选好编号后，都追加一次 |

如果误把 append 放进 else，只有未知字符的编号会被保存。如果放到整个 for 外面，它只会在循环结束后追加最后一个编号。

空行帮助阅读，但不会结束代码块；代码块范围由缩进决定。

## 6. 未知标记带来的信息损失

使用本课固定字典：

| 原文 | 编号 |
| --- | --- |
| 我爱大模型 | `[0, 1, 6, 4, 5]` |
| 我爱小模型 | `[0, 1, 6, 4, 5]` |

“大”和“小”都变成 6。仅凭编号，无法确定原来的字符是哪一个。按编号还原时，也只能在这个位置放回未知标记。

这是本课手工字符编码器的规则。真实 tokenizer 的编码方法与未知字符策略取决于具体实现，不能把这个例子的限制直接推广给所有模型。


## 7. 完整程序

下一单元与配套 `lesson_02_dict_and_conditions.py` 的内容一致，可独立运行。

在脚本所在目录的终端输入 `python lesson_02_dict_and_conditions.py`；使用 py 或 python3 的系统，替换命令开头即可。


````python
# 第 02 课：用字典和条件判断处理未知字符。
# 运行方法：python lesson_02_dict_and_conditions.py
# 只需要 Python；没有第三方依赖，不需要显卡或网络。
# <unk> 是本程序约定的未知字符标记，不是 Python 自动识别的语法。

text = "我爱大模型"

# 沿用第 01 课的 0–5 号，新增 6 号表示未知字符。
token_to_id = {
    "我": 0,
    "爱": 1,
    "学": 2,
    "习": 3,
    "模": 4,
    "型": 5,
    "<unk>": 6,
}

token_ids = []

for char in text:
    if char in token_to_id:
        token_id = token_to_id[char]
    else:
        token_id = token_to_id["<unk>"]

    # 每个字符选好编号之后，都追加一次。
    token_ids.append(token_id)

print("原文：", text)
print("编号：", token_ids)
````
