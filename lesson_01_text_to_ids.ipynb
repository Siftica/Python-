## 2. 列表与从 0 开始的索引

列表用方括号 `[]` 保存一组按顺序排列的值。本课把字符列表叫作 `vocab`，意思是词表。

| 索引 | 字符 |
| --- | --- |
| 0 | 我 |
| 1 | 爱 |
| 2 | 学 |
| 3 | 习 |
| 4 | 模 |
| 5 | 型 |

Python 列表的第一个位置编号是 **0**。`vocab[4]` 取索引 4 的值，也就是第 5 个字符。

**先预测：** 下面两次 print 分别显示哪个字符？


```python
vocab = ["我", "爱", "学", "习", "模", "型"]
print(vocab[0])
print(vocab[4])
```

## 3. for 循环：逐个处理字符

`for char in text:` 会按顺序取出 text 中的字符，每次把当前字符赋给变量 char。

冒号后面的缩进行属于循环。本课统一用 **4 个空格**缩进。下面的 print 会执行四次。

**先预测：** 输出是一行完整的句子，还是四行字符？


```python
text = "我爱学习"
for char in text:
    print(char)
```

## 4. 用 append 逐步保存结果

`token_ids = []` 创建空列表。`token_ids.append(0)` 把数字 0 加到列表末尾。

**先预测：** 依次追加 0 和 1 后，列表里有什么？


```python
token_ids = []
token_ids.append(0)
token_ids.append(1)
print(token_ids)
```

### 4.1 查编号并保存

`vocab.index(char)` 查找字符在列表里的位置。比如查“我”得到 0，查“学”得到 2。

下面把查找操作放进循环。循环里的 print 用来观察当前字符和查到的编号；循环外的 print 显示最终列表。

每次运行时，都先把 token_ids 重新设为空列表，所以重复运行不会不断追加旧结果。


```python
text = "我爱学习"
vocab = ["我", "爱", "学", "习", "模", "型"]
token_ids = []

for char in text:
    token_id = vocab.index(char)
    token_ids.append(token_id)
    print("当前字符与编号：", char, token_id)

print("完整编号：", token_ids)
```

编号由这份词表的位置决定。改变词表顺序，编号就可能改变。编号 4 并不表示“模”比编号 0 的“我”更重要。

## 5. 根据编号还原文字

`vocab[token_id]` 根据编号取出字符。先准备空字符串 `restored = ""`，再把取出的字符依次接到后面。

在 `restored = restored + vocab[token_id]` 中，先计算等号右侧，再把新字符串赋回 restored。

下面单元自带需要的词表和编号，可以独立运行。


```python
vocab = ["我", "爱", "学", "习", "模", "型"]
token_ids = [0, 1, 2, 3]
restored = ""

for token_id in token_ids:
    restored = restored + vocab[token_id]

print("还原：", restored)
```

## 6. 完整程序

下一单元与配套 `lesson_01_text_to_ids.py` 的内容一致。它包含全部变量定义，可单独运行。

运行脚本时，在脚本所在目录的终端输入 `python lesson_01_text_to_ids.py`；若你使用 `py` 或 `python3`，把开头的 python 换成对应命令。

终端命令不需要粘贴到 Notebook 的 Python 代码单元中。


```python
# 第 01 课：把中文变成编号，再按编号还原。
# 运行方法：python lesson_01_text_to_ids.py
# 只需要 Python；没有第三方依赖，不需要显卡或网络。
# 本课的输入字符需要已经存在于 vocab 中；未知字符在第 02 课处理。

# 1. 准备原文和固定词表。
text = "我爱学习"
vocab = ["我", "爱", "学", "习", "模", "型"]

# 2. 编码：按顺序查找每个字符的位置，保存编号。
token_ids = []
for char in text:
    token_id = vocab.index(char)
    token_ids.append(token_id)

print("原文：", text)
print("词表：", vocab)
print("编号：", token_ids)

# 3. 还原：按编号查出字符，再依次拼接。
restored = ""
for token_id in token_ids:
    restored = restored + vocab[token_id]

print("还原：", restored)
```

## 8. 常见问题

| 现象 | 先检查什么 |
| --- | --- |
| NameError | 是否先运行了定义变量的单元；变量名是否拼写一致 |
| 缩进相关错误 | for 末尾是否有冒号；循环里的语句是否统一缩进 4 个空格 |
| IndexError | 是否使用了列表里不存在的位置；当前词表只有 0–5 号位置 |
| ValueError | `vocab.index(...)` 要找的字符是否在词表中 |

例如，`vocab.index("大")` 会找不到“大”。这条报错示例只写在文字中，不会中断“运行全部单元”。第 02 课会学习如何处理未知字符。
