## 1. 为什么把代码写成函数？

前两课已经有了编码循环。如果要处理十段文字，我们希望复用同一段逻辑：每次传入不同文字，得到这次的编号结果。

函数可以为一段处理步骤起名字，并约定它接收什么输入、返回什么结果。后面处理训练数据、调用模型和实现工具时，都会继续使用这种组织方式。

先从一个小函数开始：给传入的文字加上“原文：”这个前缀。


````python
def add_label(text):
    return "原文：" + text
````

### 1.1 看懂函数定义

```python
def add_label(text):
    return "原文：" + text
```

| 部分 | 含义 |
| --- | --- |
| `def` | 开始定义一个函数 |
| `add_label` | 我们给函数起的名字 |
| `(text)` | 声明一个接收输入的参数 |
| `:` 与缩进 | 标出函数体的范围 |
| `return ...` | 把计算结果交给调用处，并结束这次函数调用 |

运行上面的定义单元，只是让 Python 记住这个函数；函数体会在调用时执行。

**先预测：** 下一单元会打印什么？


````python
message = add_label("我爱学习")
print(message)
````

## 2. 参数、实参与返回值

在定义 `add_label(text)` 中，text 是**参数**，它是函数接收输入时使用的名字。

在调用 `add_label("我爱学习")` 中，字符串“我爱学习”是这次传入的**实参**。

执行 `message = add_label("我爱学习")` 时：

1. 把“我爱学习”交给这次调用里的参数 text。
2. 执行函数体，计算出“原文：我爱学习”。
3. return 把这个字符串交回调用处。
4. 调用处把返回的字符串赋给 message。

然后 `print(message)` 才把结果显示出来。改变实参，可以复用同一个函数。


````python
another_message = add_label("我爱模型")
print(another_message)
````

## 3. return 与 print 的区别

`return` 把结果交给调用处，后面的代码可以继续使用它。`print` 把内容显示出来。

下面这个函数只有 print，没有用 return 提供计算结果。先预测它会打印几行，以及 shown 最后得到什么。


````python
def show_label(text):
    print("原文：" + text)

shown = show_label("我爱学习")
print("调用得到的返回值：", shown)
````

第一行由函数里面的 print 显示。第二行显示 `None`：函数执行到末尾而没有用 return 返回一个值时，调用的返回值就是 None。

None 不是字符串“None”，也不是空字符串；这里用它表示函数没有提供我们想要的处理结果。

先前的 add_label 返回字符串，所以后面的代码可以接着使用它：


````python
message = add_label("我爱学习")
final_message = message + "。"
print(final_message)
````

Notebook 可能自动展示单元最后一个表达式的值。因此，仅凭屏幕上有输出，不能判断函数内部是否调用了 print。本课统一先用变量接收返回值，再显式 print，方便观察两步的区别。

## 4. 把编码循环整理成 encode 函数

下面的循环与你上一课写过的一样。现在做三件事：

- 把它放进 `def encode(text, token_to_id):`。
- 从参数接收输入文字和字典。
- 用 `return token_ids` 把完整列表交给调用处。

这个函数需要两个参数；调用时也按顺序提供两个实参。


````python
def encode(text, token_to_id):
    token_ids = []

    for char in text:
        if char in token_to_id:
            token_id = token_to_id[char]
        else:
            token_id = token_to_id["<unk>"]

        token_ids.append(token_id)

    return token_ids

token_to_id = {
    "我": 0,
    "爱": 1,
    "学": 2,
    "习": 3,
    "模": 4,
    "型": 5,
    "<unk>": 6,
}

ids = encode("我爱模型", token_to_id)
print(ids)
````

### 4.1 为什么每次都创建空列表？

`token_ids = []` 在函数内部。每次调用 encode，都会新建一个结果列表，再收集这次输入的编号。

这个变量属于函数本次调用的局部范围。调用处用自己的变量名接收返回结果，名字不必也叫 token_ids。

**先预测：** 连续调用两次，second_ids 会不会包含第一次的结果？


````python
first_ids = encode("我爱", token_to_id)
second_ids = encode("学习", token_to_id)

print("第一次：", first_ids)
print("第二次：", second_ids)
````

### 4.2 return 的缩进决定什么时候结束

在 encode 中，`return token_ids` 位于 **函数内部、for 循环外部**：所有字符都处理完，再返回整个列表。

```python
def encode(text, token_to_id):
    token_ids = []
    for char in text:
        # 在这里处理当前字符并追加编号
        ...
    return token_ids
```

上面是结构示意，省略号表示这里省略了已经讲过的循环逻辑；它不是完整程序，不需要复制运行。

如果误把 return 放到循环内部，本例会处理完第一个字符就结束函数。空输入时，循环一次也不会执行，更无法从循环内部返回空列表。

### 4.3 把依赖通过参数传入

encode 直接使用传进来的 token_to_id。换一份词表时，传入另一份字典即可复用函数。函数本身不会修改这份词表。

传入的词表需要包含 `"<unk>"`。本课保留这个输入约定，异常处理在后续学习。


````python
small_vocab = {"你": 0, "好": 1, "<unk>": 2}
small_ids = encode("你好啊", small_vocab)
print(small_ids)
````

## 5. 写 decode：把编号转回字符串

第 01 课已经用列表按编号取字符。这里把同样的逻辑整理成 decode 函数。

本课的 id_to_token 是列表，其位置必须与 token_to_id 中的编号一致：

| 编号 | id_to_token 中对应的内容 |
| --- | --- |
| 0 | 我 |
| 1 | 爱 |
| 2 | 学 |
| 3 | 习 |
| 4 | 模 |
| 5 | 型 |
| 6 | `<unk>` |

decode 的两个参数是编号列表与对应字符表。下面单元自带所需定义，可以独立运行。


````python
def decode(token_ids, id_to_token):
    restored = ""

    for token_id in token_ids:
        restored = restored + id_to_token[token_id]

    return restored

id_to_token = ["我", "爱", "学", "习", "模", "型", "<unk>"]

restored = decode([0, 1, 4, 5], id_to_token)
print(restored)
````

## 6. 组合两个函数

一个函数的返回值可以作为另一个函数的输入。

先用 encode 得到编号列表，再把这份列表交给 decode。

**先预测：** 输入里有“大”，解码结果会完整恢复原文吗？


````python
text = "我爱大模型"
ids = encode(text, token_to_id)
restored = decode(ids, id_to_token)

print("原文：", text)
print("编号：", ids)
print("解码：", restored)
````

“大”在编码时变成了未知字符编号 6，解码只能得到对应标记 `<unk>`。函数执行正确，也不能恢复已经丢失的信息。

这里继续使用手工固定字符表，没有模型训练；未知标记是 else 分支按约定加入的。

## 7. 空输入也是一种需要想清楚的情况

| 调用 | 返回值 | 类型 |
| --- | --- | --- |
| `encode("", token_to_id)` | `[]` | 空列表 |
| `decode([], id_to_token)` | `""` | 空字符串 |

没有要处理的元素时，循环执行零次；随后函数仍会执行循环后面的 return。


````python
empty_ids = encode("", token_to_id)
empty_text = decode(empty_ids, id_to_token)

print("空文字的编号：", empty_ids)
print("空编号的解码：", empty_text)
````

第二行冒号后没有文字，表示返回的是空字符串。

## 8. 完整程序

下面单元与配套 `lesson_03_functions_and_returns.py` 完全一致，包含两个函数、两份对应表和三种输入示例，可以独立运行。

这里的 decode 只接收与对应字符表匹配的有效编号，例如这份 encode 产生的编号。更换词表时，需要同步更换解码表；本课暂不处理外部传入的非法编号。

在脚本所在目录的终端运行 `python lesson_03_functions_and_returns.py`。如果你此前使用 py 或 python3，沿用对应命令。


````python
# 第 03 课：函数、参数与返回值。
# 运行方法：python lesson_03_functions_and_returns.py
# 本程序只使用 Python 标准功能，无需第三方包、模型文件或显卡。
# encode 接收文字与词表，返回编号列表。
# decode 接收有效编号与对应字符表，返回字符串。
# 两份表的编号必须一致；词表需包含 <unk>。


def encode(text, token_to_id):
    token_ids = []

    for char in text:
        if char in token_to_id:
            token_id = token_to_id[char]
        else:
            token_id = token_to_id["<unk>"]

        token_ids.append(token_id)

    return token_ids


def decode(token_ids, id_to_token):
    restored = ""

    for token_id in token_ids:
        restored = restored + id_to_token[token_id]

    return restored


# 保留前两课的编号；大、小等未收录的字符仍使用 6 号。
token_to_id = {
    "我": 0,
    "爱": 1,
    "学": 2,
    "习": 3,
    "模": 4,
    "型": 5,
    "<unk>": 6,
}
id_to_token = ["我", "爱", "学", "习", "模", "型", "<unk>"]

# 用同一组函数处理三种输入：已知字符、含未知字符、空字符串。
texts = ["我爱学习", "我爱大模型", ""]

for text in texts:
    token_ids = encode(text, token_to_id)
    restored = decode(token_ids, id_to_token)

    print("原文：", text)
    print("编号：", token_ids)
    print("解码：", restored)
    print()
````

## 10. 常见问题

| 现象 | 先检查什么 |
| --- | --- |
| 调用函数时 NameError | 是否先执行了定义函数的单元；函数名是否一致 |
| TypeError 提示缺少参数 | 调用 encode 或 decode 时，是否提供了两个实参 |
| 屏幕上看到了编号，变量却是 None | 是否只写了 print，忘记用 return 返回结果 |
| 只返回第一个字符的编号 | return 是否被缩进到 for 循环里面 |
| 访问函数里的变量时报错 | 是否应该接收函数的返回值，而不是直接访问局部变量 |
| 空输入返回 None | return 是否错误地放进循环或分支里 |
| 解码文字对不上编号 | 编码字典和解码列表是否使用同一套编号 |

### 练习 C 参考

对已有字符，使用一致的两份表可以还原。未知字符被统一映射后，只能得到未知标记。


````python
def encode(text, token_to_id):
    token_ids = []

    for char in text:
        if char in token_to_id:
            token_id = token_to_id[char]
        else:
            token_id = token_to_id["<unk>"]

        token_ids.append(token_id)

    return token_ids

def decode_c(token_ids, id_to_token):
    restored = ""

    for token_id in token_ids:
        restored = restored + id_to_token[token_id]

    return restored

token_to_id = {
    "我": 0,
    "爱": 1,
    "学": 2,
    "习": 3,
    "模": 4,
    "型": 5,
    "<unk>": 6,
}
id_to_token = ["我", "爱", "学", "习", "模", "型", "<unk>"]
examples = ["学习模型", "我爱大模型", ""]
for example in examples:
    ids = encode(example, token_to_id)
    decoded = decode_c(ids, id_to_token)
    print("输入：", example)
    print("编号：", ids)
    print("解码：", decoded)
    print()
````
